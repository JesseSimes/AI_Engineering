```python
arr1=np.arange(1,6)
arr2=np.arange(6,11)
print(arr1)
print(arr2)
```

```python
#operations
print(arr1 + arr2)
print(arr1 - arr2)
print(arr1 * arr2)
print(arr1 / arr2)
print(arr1 % arr2)
print(arr1 // arr2)
print(arr1 ** arr2)

#broadcasting
print(arr1 + 10)
print(arr2 + 10)
```
